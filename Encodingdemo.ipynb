## Examples for Encoding

```python
# sklearn.preprocessing is imported because encoding is a data preprocessing step. It provides tools to convert
# categorical (text) data into numerical data so machine learning algorithms can understand and process it.
# sklearn stands for Scikit-learn, a popular Python library used for Machine Learning.
# preprocessing means preparing the data before giving it to a machine learning model.
```

```python
# Label Encoding
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder,OrdinalEncoder
```

```python
# Create the DataFrame
# A DataFrame is a two-dimensional table-like data structure in Python (provided by the Pandas library)
# that stores data in rows and columns. Dataframe is a constructor df is the object created, a dictionary 
# is created gender(key) is the column, values are male, female
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male','Female']
})
```

```python
# Create LabelEncoder object
# Create a LabelEncoder object and store it in the variable encoder so it can be used to 
# convert categorical (text) data into numerical values.
encoder = LabelEncoder()
```

```python
# Apply Label Encoding
# encoder → is an object (instance) of the LabelEncoder class.
# fit_transform() → is a method of the LabelEncoder class.
# df['Gender'] is a pandas series
df['Gender'] = encoder.fit_transform(df['Gender'])
```

```python
#The LabelEncoder converts text values into numbers. Since it assigns numbers in alphabetical order, 
#Female becomes 0 and Male becomes 1. Every occurrence of 'Female' is replaced with 0, and every occurrence
#of 'Male' is replaced with 1.
print(df)
```

```python
# OneHotEncoding
```

```python
# Create the DataFrame
df1 = pd.DataFrame({
    'Department': ['IT', 'HR', 'Finance', 'IT', 'HR']
})
```

```python
#from sklearn.preprocessing import OneHotEncoder
```

```python
# Create OneHotEncoder object
# sparse_output = false--> tells OneHotEncoder how to display the encoded output.
encoder = OneHotEncoder(sparse_output=False, dtype=int)
```

```python
# Apply One-Hot Encoding
encoded = encoder.fit_transform(df1[['Department']])
```

```python
# Convert the encoded array into a DataFrame
# columns= specifies the column names for the new DataFrame.
# encoder.get_feature_names_out(['Department'])-->This is a method of the OneHotEncoder object.
# It automatically generates meaningful column names.
encoded_df1 = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['Department'])
)
```

```python
#OneHotEncoder() creates a separate column for each department.
#A value of 1 indicates the employee belongs to that department, 
#while 0 indicates they do not.
print(encoded_df1)
```

```python
# Ordinal Encoding
# Create the DataFrame
df2 = pd.DataFrame({
    'Customer_Satisfaction': ['Poor', 'Good', 'Excellent', 'Average', 'Good', 'Poor']
})
```

```python
# Create OrdinalEncoder object
# it creates an OrdinalEncoder object and tells it the exact order of the categories to use during encoding.
# it expects the value as 2 dimensional array that's why 2 [[]]
encoder = OrdinalEncoder(categories=[['Poor', 'Average', 'Good', 'Excellent']]) 
```

```python
# Apply Ordinal Encoding
#encoder → Object of the OrdinalEncoder class.
#fit() → Learns the order you specified:
# transform() → Converts the categories into numbers.
df2['Customer_Satisfaction'] = encoder.fit_transform(df2[['Customer_Satisfaction']]).astype(int)
```

```python
print(df2)
```
